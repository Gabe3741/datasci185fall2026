**(b)** Imagine flipping a coin repeatedly until we get 10 heads (in total, not in a row). How many tails will you see before seeing the 10th head?

The **negative binomial distribution** describes exactly this. Given

- `n` = the number of heads needed to stop the experiment, and
- `p` = the probability of heads on a single flip,

it tells us the probability of getting each possible number of tails before the experiment ends (0 tails, 1 tail, 2 tails, ... ).

<font size="4"> 

The probability of exactly `k` tails before seeing `n` heads is

$$P(k \text{ tails}) = \frac{(k+n-1)!}{k!\,(n-1)!}\, p^{k}\,(1-p)^{n}$$

Use the scipy function `scipy.stats.nbinom` to plot the probability mass function (pmf) over the range $k=0, 1, 2, ..., 9, 10$ when we stop after flipping $n=10$ heads of a fair coin (so $p = 0.5$).

```python
k = range(0,10)
print(k)

help(plt.xscale)
help(stats.nbinom)
```

```python
# your code here
import scipy.stats as stats
import matplotlib.pyplot as plt


k = range(0,10)
n = 10
p = .5

dist = stats.nbinom.pmf(k,n,p)
print(dist)

plt.scatter(x = k,y = dist)
plt.show()
```

```python
import numpy as np 
import scipy.stats as stats 

# your code here
# create and show BOTH plots in this cell

x = np.linspace(-5,5,100)

LinspaceDist = stats.norm.pdf(x)
plt.plot(x,LinspaceDist)
plt.xlabel("100 points between -5 and 5")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Probability between points -5 to 5")
plt.show()

LinspaceDist2 = stats.norm.pdf(x,loc = 1,scale = 2)
plt.plot(x,LinspaceDist2)
plt.xlabel("100 points between -5 and 5")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Probability between points -5 to 5 \n stdev = 2, mean = 1")
plt.show()
```
